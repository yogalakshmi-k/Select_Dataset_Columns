```python
import pandas as pd

data = {
    "Name": ["Kavya", "Arun", "Meena", "Vijay", "Divya"],
    "Marks": [90, 76, 85, 88, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.loc[:, "Name"])
print(df.loc[:, ["Name", "Marks"]])
```

```python
print("Select First Row:")
print(df.iloc[0])

print("\nSelect First Column:")
print(df.iloc[:, 0])

print("\nSelect First Two Columns:")
print(df.iloc[:, 0:2])

print("\nSelect First Three Rows:")
print(df.iloc[0:3])
```

```python
print("Selected Students and Marks:")
print(df.iloc[1:4, [0, 1]])

print("\nStudents Scoring Above 85:")
print(df[df["Marks"] > 85])
```
